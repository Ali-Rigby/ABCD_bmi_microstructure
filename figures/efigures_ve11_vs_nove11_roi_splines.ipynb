### ROI spline plots — VE11 vs noVE11 (matched 6.0)

Loads **combined-sex** s9 ROI-averaged splines (not sex-stratified) from:

`ve11_vs_noVE11_6_0/{bmip|perc_median}/{VE11|noVE11}/results/<run>/roi_data_for_plots/`

**Knobs (next cell):**
- `BMI_PHENO_VARIABLE` — `bmip` or `perc_median`
- `BMI_IMAGING_RESULTS_RUN` — default `results_2026_09_02` (bmip) / `results_2026_10_02` (perc_median)
- `ARM_A` / `ARM_B` — which arms to overlay (default VE11 vs noVE11; swap to flip)
- `correction_type` — default `fdr` (also `all_voxels`, `bonferroni`)

**Prerequisite:** `bash scripts/sensitivity/ve11_vs_nove11/run_s9.sh`


```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
%matplotlib inline


# --- VE11 vs noVE11 (matched 6.0) knobs (edit here) ---
os.environ["BMI_PHENO_VARIABLE"] = "perc_median"  # "bmip" or "perc_median"
os.environ["BMI_IMAGING_CEPH_ROOT"] = "/space/ceph/1/ABCD/users/arigby/2026_bmi_imaging_analysis"
# Leave unset to use pheno-specific defaults below, or override:
# os.environ["BMI_IMAGING_RESULTS_RUN"] = "results_2026_09_02"

CEPH_ROOT = os.environ.get("BMI_IMAGING_CEPH_ROOT", "/space/ceph/1/ABCD/users/arigby/2026_bmi_imaging_analysis")
PHENO = os.environ.get("BMI_PHENO_VARIABLE", "perc_median").strip()
ANALYSIS_ROOT = "ve11_vs_noVE11_6_0"

DEFAULT_RESULTS_RUN = {
    "bmip": "results_2026_09_02",
    "perc_median": "results_2026_10_02",
}
RESULTS_RUN = os.environ.get(
    "BMI_IMAGING_RESULTS_RUN",
    DEFAULT_RESULTS_RUN.get(PHENO, "results_2026_09_02"),
).strip()

# Which combined-sex arms to overlay (not sex-stratified)
ARM_A = "VE11"     # first series / left panel
ARM_B = "noVE11"   # second series / right panel
# Swap to flip: ARM_A, ARM_B = "noVE11", "VE11"
if {ARM_A, ARM_B} != {"VE11", "noVE11"} or ARM_A == ARM_B:
    raise ValueError(f"ARM_A/ARM_B must be a permutation of VE11 and noVE11 (got {ARM_A!r}, {ARM_B!r})")

PHENO_PROFILES = {
    "bmip": {
        "xname": "bmi",
        "design_tag": "Sbmi",
        "pheno_xlabel": "BMI percentile",
        "vector_file": "bmi_percentile_vector.txt",
        "x_plot_min": 0,
        "x_plot_max": 100,
        "x_from_data": False,
    },
    "perc_median": {
        "xname": "perc_median",
        "design_tag": "Sperc",
        "pheno_xlabel": "Percent from median",
        "vector_file": "perc_median_vector.txt",
        "x_plot_min": None,  # filled from sample min/max
        "x_plot_max": None,
        "x_from_data": True,  # floor(min), ceil(max) -> integers
    },
}
if PHENO not in PHENO_PROFILES:
    raise ValueError(f"BMI_PHENO_VARIABLE must be bmip or perc_median (got {PHENO!r})")
prof = PHENO_PROFILES[PHENO]
xname = prof["xname"]
design_tag = prof["design_tag"]
pheno_xlabel = prof["pheno_xlabel"]

import math
import numpy as np

def perc_median_integer_xlim(vector_path: str) -> tuple[int, int]:
    """Outward integer bounds from sample perc_median values (floor min, ceil max)."""
    vals = pd.read_csv(vector_path, sep=r"\s+", header=None, engine="python").iloc[:, 0].astype(float)
    vals = vals[np.isfinite(vals)]
    if vals.empty:
        raise ValueError(f"No finite values in {vector_path}")
    xmin = int(math.floor(float(vals.min())))
    xmax = int(math.ceil(float(vals.max())))
    if xmin == xmax:
        xmax = xmin + 1
    return xmin, xmax

# Env overrides win if set; else bmip uses fixed 0-100; perc_median uses data range
_env_xmin = os.environ.get("BMI_IMAGING_PLOT_XMIN", "").strip()
_env_xmax = os.environ.get("BMI_IMAGING_PLOT_XMAX", "").strip()
if _env_xmin and _env_xmax:
    x_plot_min = int(_env_xmin)
    x_plot_max = int(_env_xmax)
elif prof["x_from_data"]:
    _vec = os.path.join(CEPH_ROOT, ANALYSIS_ROOT, PHENO, "datafiles", prof["vector_file"])
    if not os.path.isfile(_vec):
        raise FileNotFoundError(
            f"Need {prof['vector_file']} for perc_median x-limits:\n  {_vec}"
        )
    x_plot_min, x_plot_max = perc_median_integer_xlim(_vec)
    print(f"perc_median x-limits from {_vec}: data -> [{x_plot_min}, {x_plot_max}]")
else:
    x_plot_min = int(prof["x_plot_min"])
    x_plot_max = int(prof["x_plot_max"])

modality = "rni"
effect_type = "splines"
effect_types = [effect_type]  # loops below expect a list
xnames = [xname]
correction_type = "fdr"  # 'bonferroni', 'fdr', or 'all_voxels'

# Shared plot root for the VE11 vs noVE11 comparison
plots_base_directory = os.path.join(
    CEPH_ROOT, ANALYSIS_ROOT, PHENO, "plots_ve11_noVE11", RESULTS_RUN
)


def find_arm_roi_dir(arm: str) -> str:
    """Locate designmat1_* ROI export folder for a scanner arm (combined-sex)."""
    base = os.path.join(
        CEPH_ROOT, ANALYSIS_ROOT, PHENO, arm, "results", RESULTS_RUN, "roi_data_for_plots"
    )
    if not os.path.isdir(base):
        raise FileNotFoundError(
            f"Missing ROI spline export root:\n  {base}\n"
            f"Run s9 for {PHENO}/{arm} first "
            f"(bash scripts/sensitivity/ve11_vs_nove11/run_s9.sh)."
        )
    # Arm is selected by path; designmat names use Scanner (VE11) vs ScanSoft (noVE11)
    # and do not embed the arm label.
    matches = sorted(glob.glob(os.path.join(base, f"designmat1_*{design_tag}*")))
    matches = [m for m in matches if os.path.basename(m).startswith("designmat1_")]
    # Prefer combined-sex (exclude designmat1f_/1m_ if ever present)
    matches = [m for m in matches if not os.path.basename(m).startswith(("designmat1f_", "designmat1m_"))]
    if not matches:
        matches = sorted(glob.glob(os.path.join(base, "designmat*")))
    if not matches:
        raise FileNotFoundError(f"No designmat* under {base}")
    return matches[0]


arm_dirs = {
    "VE11": find_arm_roi_dir("VE11"),
    "noVE11": find_arm_roi_dir("noVE11"),
}


print(f"Analysis: {ANALYSIS_ROOT}")
print(f"Pheno: {PHENO} (xname={xname})")
print(f"Results run: {RESULTS_RUN}")
print(f"{ARM_A} ROI dir: {arm_dirs[ARM_A]}")
print(f"{ARM_B} ROI dir: {arm_dirs[ARM_B]}")
print(f"Overlay: {ARM_A} vs {ARM_B}")
print(f"correction_type: {correction_type}")
print(f"Plot x range: [{x_plot_min}, {x_plot_max}]")
print(f"Plots out: {plots_base_directory}")


# Colors keyed by scanner arm (combined-sex models)
colors = {"VE11": "#2A9D8F", "noVE11": "#E76F51"}
ci_colors = {"VE11": "#A8DAD5", "noVE11": "#F4A261"}

custom_roi_colors = {
    'Bilateral-Thalamus-Proper': '#fc8d62',
    'Bilateral-Caudate': '#e78ac3',
    'Bilateral-Putamen': '#8da0cb',
    'Bilateral-Accumbens-area': '#a6d854',
    'Bilateral-Pallidum': '#66c2a5',
    'Bilateral-Hippocampus': '#ffff33',          # Yellow
    'Bilateral-Amygdala': '#a65628',             # Brown
    'Bilateral-VentralDC': '#f781bf',            # Pink
}

def load_data(directory, effect_type, xname, laterality_type='lateralized', correction_type='fdr'):
    """
    Load data from the new hierarchical directory structure.
    
    Parameters:
    -----------
    directory : str
        Base directory containing the design matrix results
    effect_type : str
        Type of effect (e.g., 'splines')
    xname : str
        Variable name (e.g., pheno xname from s9)
    laterality_type : str
        Type of laterality: 'lateralized', 'bilateral', or 'non_lateralized'
    correction_type : str
        Multiple comparison correction: 'bonferroni', 'fdr', or 'all_voxels'
    
    Returns:
    --------
    dict : Dictionary mapping ROI names to DataFrames
    """
    # Navigate to betas/laterality_type/correction_type subdirectory
    correction_dir = os.path.join(directory, 'betas', laterality_type, correction_type)
    if not os.path.exists(correction_dir):
        print(f"Warning: Directory not found: {correction_dir}")
        return {}
    
    pattern = f"{modality}_{effect_type}_{xname}_CI_"
    files = [f for f in os.listdir(correction_dir) if f.startswith(pattern) and f.endswith('.txt')]
    
    print(f"\n{laterality_type} - Looking in: {correction_dir}")
    print(f"Pattern: {pattern}")
    print(f"Found {len(files)} files: {files}")
    
    data = {}
    for file in files:
        file_path = os.path.join(correction_dir, file)
        # Extract ROI name (everything after 'CI_' and before '.txt')
        roi_name = file.replace(pattern, '').replace('.txt', '')
        try:
            df = pd.read_csv(file_path)
            data[roi_name] = df
        except Exception as e:
            print(f"Error loading {file_path}: {e}")
    
    print(f"Loaded {len(data)} ROIs: {list(data.keys())}")
    return data

# Function to sort ROIs with left and right paired first, then unpaired
PAULI_ROIS = [
    "Putamen",
    "Caudate",
    "Nucleus Accumbens",
    "Extended Amygdala",
    "Globus Pallidus external",
    "Globus Pallidus internal",
    "Substantia Nigra pars compacta",
    "Red Nucleus",
    "Substantia Nigra pars reticulata",
    "Parabrachial Pigmented Nucleus",
    "Ventral Tegmental Area",
    "Ventral Pallidum",
    "Habenular Nuclei",
    "Hypothalamus",
    "Mammillary Nucleus",
    "Subthalamic Nucleus",
]

OTHER_NON_LATERALIZED_ROIS = ["CC_Anterior", "Fmaj", "Fmin"]


roi_display_names = {
    "Fmaj": "Forceps major",
    "Fmin": "Forceps minor",
    "Accumbens-area": "Nucleus accumbens",
    "Thalamus-Proper": "Thalamus",
}

def display_name(roi_name):
    return roi_display_names.get(roi_name, roi_name)

def filter_rois(data, roi_names):
    """Return subset of loaded ROI dict for requested names (preserving order)."""
    return {roi: data[roi] for roi in roi_names if roi in data}

def sort_rois(rois):
    paired_rois = []
    # Known non-lateralized ROIs (fiber/CC + Pauli); only include if present
    known_single_rois = [
        roi for roi in (OTHER_NON_LATERALIZED_ROIS + PAULI_ROIS) if roi in rois
    ]
    single_rois = known_single_rois
    
    # Group left-right pairs together (handle both "Left"/"Right" and "LH-"/"RH-" naming conventions)
    left_rois = sorted([roi for roi in rois if roi.startswith("Left") or roi.startswith("LH-")])
    for left_roi in left_rois:
        if left_roi.startswith("Left"):
            right_roi = left_roi.replace("Left", "Right")
        elif left_roi.startswith("LH-"):
            right_roi = left_roi.replace("LH-", "RH-")
        else:
            continue
            
        if right_roi in rois:
            paired_rois.extend([left_roi, right_roi])
    
    # Remaining single ROIs (excluding the known non-lateralized ones we already handled)
    remaining_rois = [roi for roi in rois if roi not in paired_rois and roi not in single_rois]
    return paired_rois + remaining_rois + single_rois

# Plot data for each effect type and xname

def safe_filename_suffix(text):
    """Make title suffix safe for PDF/PNG filenames."""
    s = text.replace(': ', '').replace(' ', '_')
    for ch in '/\\:*?"<>|':
        s = s.replace(ch, '_')
    while '__' in s:
        s = s.replace('__', '_')
    return s.strip('_')

def roi_ylim(data_a, data_b, roi_name, x_min, x_max):
    """Y-axis limits for one ROI from CI bounds within the x range."""
    y_min, y_max = float('inf'), float('-inf')
    for data in (data_a, data_b):
        if roi_name not in data:
            continue
        df = data[roi_name]
        df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
        if df.empty:
            continue
        y_min = min(y_min, df['CI_lower'].min())
        y_max = max(y_max, df['CI_upper'].max())
    if y_min == float('inf'):
        return None
    return y_min, y_max


def plot_data(data_a, data_b, xname, effect_type, laterality_type='lateralized', title_suffix="", save_pdf=False, save_png=False, share_y_within_pair=None, file_tag=None):
    # Check if there's any data to plot
    if not data_a and not data_b:
        print(f"No data found for {laterality_type} ROIs. Skipping plot.")
        return
    
    # Get all unique ROIs from both arms
    all_rois = set(list(data_a.keys()) + list(data_b.keys()))
    sorted_rois = sort_rois(all_rois)
    n_rois = len(sorted_rois)
    
    if n_rois == 0:
        print(f"No ROIs found for {laterality_type}. Skipping plot.")
        return
    
    fig, axes = plt.subplots(nrows=(n_rois + 1) // 2, ncols=2, figsize=(15, (n_rois + 1) * 2.5))

    # font size for the main title
    fig.suptitle(f"{modality.upper()} {effect_type.capitalize()} Effect for {xname.capitalize()} ({laterality_type}) {title_suffix}", fontsize=24)
    axes = axes.flatten()

    # Define the x-axis range limits
    x_min, x_max = x_plot_min, x_plot_max

    # Prepare PNG directory if saving individual ROI plots
    if save_png:
        png_dir = os.path.join(plots_base_directory, 'betas', laterality_type, correction_type, modality, 'pngs')
        os.makedirs(png_dir, exist_ok=True)

    if share_y_within_pair is None:
        share_y_within_pair = laterality_type == 'lateralized'

    for i in range(0, len(sorted_rois), 2):
        roi_pair = sorted_rois[i:i+2]  # Take pairs of ROIs (e.g., Left and Right)

        # Lateralized L/R panels share y; non-lateralized panels use per-ROI limits
        if share_y_within_pair:
            y_min, y_max = float('inf'), float('-inf')
            for roi_name in roi_pair:
                limits = roi_ylim(data_a, data_b, roi_name, x_min, x_max)
                if limits is None:
                    continue
                roi_y_min, roi_y_max = limits
                y_min = min(y_min, roi_y_min)
                y_max = max(y_max, roi_y_max)
            if y_min == float('inf') or y_max == float('-inf'):
                print(f"Warning: No data found for ROI pair {roi_pair}. Skipping.")
                continue
        else:
            has_any = any(
                roi_ylim(data_a, data_b, roi_name, x_min, x_max) is not None
                for roi_name in roi_pair
            )
            if not has_any:
                print(f"Warning: No data found for ROI pair {roi_pair}. Skipping.")
                continue

        for j, roi_name in enumerate(roi_pair):
            ax = axes[i + j]

            # Plot ARM_A within the x-axis range
            if roi_name in data_a:
                df_a = data_a[roi_name]
                df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                ax.plot(df_a['xvec'], df_a['mean'], color=colors[ARM_A], label=ARM_A)
                ax.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)

            # Plot ARM_B within the x-axis range
            if roi_name in data_b:
                df_b = data_b[roi_name]
                df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                ax.plot(df_b['xvec'], df_b['mean'], color=colors[ARM_B], label=ARM_B)
                ax.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)

            # Set axis limits and labels
            ax.set_xlim([x_min, x_max])
            if share_y_within_pair:
                ax.set_ylim([y_min, y_max])  # Shared y-axis for L/R pair
            else:
                limits = roi_ylim(data_a, data_b, roi_name, x_min, x_max)
                if limits is not None:
                    ax.set_ylim(limits)

            # Increase font size for title, labels, and legend
            ax.set_title(display_name(roi_name), fontsize=20)
            ax.set_xlabel(pheno_xlabel, fontsize=20)
            ax.set_ylabel(f'Estimated {modality.upper()}', fontsize=20)
            ax.tick_params(axis='both', which='major', labelsize=20)
            ax.legend(fontsize=18)

        # Save paired or unpaired ROIs as PNG if selected
        if save_png:
            # Check if this is a left-right pair (handle both naming conventions)
            is_paired = False
            if len(roi_pair) == 2:
                # Check for "Left"/"Right" naming
                if (roi_pair[0].startswith("Left") and roi_pair[1].startswith("Right") and
                    roi_pair[0].replace("Left", "Right") == roi_pair[1]):
                    is_paired = True
                # Check for "LH-"/"RH-" naming (thalamus ROIs)
                elif (roi_pair[0].startswith("LH-") and roi_pair[1].startswith("RH-") and
                      roi_pair[0].replace("LH-", "RH-") == roi_pair[1]):
                    is_paired = True
            
            if is_paired:
                # Create a figure with 2 subplots for the pair
                fig_individual = plt.figure(figsize=(20, 6))
                
                for j, roi_name in enumerate(roi_pair):
                    ax_individual = fig_individual.add_subplot(1, 2, j + 1)
                    
                    # Plot ARM_A
                    if roi_name in data_a:
                        df_a = data_a[roi_name]
                        df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                        ax_individual.plot(df_a['xvec'], df_a['mean'], color=colors[ARM_A], label=ARM_A)
                        ax_individual.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)
                    
                    # Plot ARM_B
                    if roi_name in data_b:
                        df_b = data_b[roi_name]
                        df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                        ax_individual.plot(df_b['xvec'], df_b['mean'], color=colors[ARM_B], label=ARM_B)
                        ax_individual.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)
                    
                    # Set same limits and styling
                    ax_individual.set_xlim([x_min, x_max])
                    ax_individual.set_ylim([y_min, y_max])
                    ax_individual.set_title(display_name(roi_name), fontsize=20)
                    ax_individual.set_xlabel(pheno_xlabel, fontsize=20)
                    ax_individual.set_ylabel(f'Estimated {modality.upper()}', fontsize=20)
                    ax_individual.tick_params(axis='both', which='major', labelsize=20)
                    ax_individual.legend(fontsize=18)
                
                # Save paired ROIs together
                base_roi_name = roi_pair[0].replace("Left", "").replace("LH-", "")
                png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_{base_roi_name}.png")
                fig_individual.tight_layout()
                fig_individual.savefig(png_filename, dpi=300, bbox_inches='tight')
                plt.close(fig_individual)
            else:
                # Save unpaired ROIs individually
                for roi_name in roi_pair:
                    fig_individual = plt.figure(figsize=(10, 6))
                    ax_individual = fig_individual.add_subplot(111)
                    
                    # Plot ARM_A
                    if roi_name in data_a:
                        df_a = data_a[roi_name]
                        df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                        ax_individual.plot(df_a['xvec'], df_a['mean'], color=colors[ARM_A], label=ARM_A)
                        ax_individual.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)
                    
                    # Plot ARM_B
                    if roi_name in data_b:
                        df_b = data_b[roi_name]
                        df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                        ax_individual.plot(df_b['xvec'], df_b['mean'], color=colors[ARM_B], label=ARM_B)
                        ax_individual.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)
                    
                    # Set limits and styling
                    ax_individual.set_xlim([x_min, x_max])
                    limits = roi_ylim(data_a, data_b, roi_name, x_min, x_max)
                    if limits is not None:
                        ax_individual.set_ylim(limits)
                    elif share_y_within_pair:
                        ax_individual.set_ylim([y_min, y_max])
                    ax_individual.set_title(display_name(roi_name), fontsize=20)
                    ax_individual.set_xlabel(pheno_xlabel, fontsize=20)
                    ax_individual.set_ylabel(f'Estimated {modality.upper()}', fontsize=20)
                    ax_individual.tick_params(axis='both', which='major', labelsize=20)
                    ax_individual.legend(fontsize=18)
                    
                    # Save as high-resolution PNG
                    png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_{roi_name}.png")
                    fig_individual.savefig(png_filename, dpi=300, bbox_inches='tight')
                    plt.close(fig_individual)

    # Remove any unused axes
    for j in range(len(sorted_rois), len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout to fit the suptitle
    plt.show()  # Display

    # Save the figure as a PDF if selected
    if save_pdf:
        # Ensure the directory for PDFs exists
        pdf_dir = os.path.join(plots_base_directory, 'betas', laterality_type, correction_type, modality, 'pdfs')
        os.makedirs(pdf_dir, exist_ok=True)

        # Update filename to include modality and correction type
        tag = f"_{file_tag}" if file_tag else ""
        pdf_filename = os.path.join(pdf_dir, f"{modality}_{effect_type}_{xname}_{safe_filename_suffix(title_suffix)}{tag}.pdf")
        with PdfPages(pdf_filename) as pdf:
            pdf.savefig(fig)
        print(f"Saved plot as {pdf_filename}")

    if save_png:
        print(f"Saved individual/paired ROI plots to {png_dir}")

# Load and plot data for the specified correction type
for effect_type in effect_types:
    for xname in xnames:
        # Plot lateralized ROIs
        data_a = load_data(arm_dirs[ARM_A], effect_type, xname, laterality_type='lateralized', correction_type=correction_type)
        data_b = load_data(arm_dirs[ARM_B], effect_type, xname, laterality_type='lateralized', correction_type=correction_type)
        plot_data(data_a, data_b, xname, effect_type, laterality_type='lateralized',
                 title_suffix=f": {correction_type.replace('_', ' ').title()}", 
                 save_pdf=True, save_png=True)
        
        # Plot non-lateralized ROIs (fiber/CC + Pauli subcortical nuclei)
        data_a = load_data(arm_dirs[ARM_A], effect_type, xname, laterality_type='non_lateralized', correction_type=correction_type)
        data_b = load_data(arm_dirs[ARM_B], effect_type, xname, laterality_type='non_lateralized', correction_type=correction_type)

        other_a = filter_rois(data_a, OTHER_NON_LATERALIZED_ROIS)
        other_b = filter_rois(data_b, OTHER_NON_LATERALIZED_ROIS)
        if other_a or other_b:
            plot_data(other_a, other_b, xname, effect_type, laterality_type='non_lateralized',
                     title_suffix=f": {correction_type.replace('_', ' ').title()}",
                     file_tag="fiber",
                     save_pdf=True, save_png=True)

        pauli_a = filter_rois(data_a, PAULI_ROIS)
        pauli_b = filter_rois(data_b, PAULI_ROIS)
        print(f"Pauli ROIs loaded: {ARM_A}={list(pauli_a.keys())}, {ARM_B}={list(pauli_b.keys())}")
        if pauli_a or pauli_b:
            plot_data(pauli_a, pauli_b, xname, effect_type, laterality_type='non_lateralized',
                     title_suffix=f": {correction_type.replace('_', ' ').title()}",
                     file_tag="pauli",
                     save_pdf=True, save_png=True)
```

```python
# Combined PNG: selected bilateral + non-lateralized ROIs (2 columns, last centered)


import matplotlib.gridspec as gridspec


def _make_combined_roi_axes(fig, n_rois=9):
    """Two columns; last ROI centered on the bottom row when n_rois is odd."""
    n_rows = (n_rois + 1) // 2
    gs = gridspec.GridSpec(n_rows, 4, figure=fig)
    axes = []
    for idx in range(n_rois):
        if idx < n_rois - 1:
            row, side = divmod(idx, 2)
            col_start = 0 if side == 0 else 2
            axes.append(fig.add_subplot(gs[row, col_start:col_start + 2]))
        else:
            axes.append(fig.add_subplot(gs[n_rows - 1, 1:3]))
    return axes

def plot_combined_selected_rois(xname, effect_type, save_png=False, save_pdf=False, save_svg=False):
    """
    Create a single combined figure with selected bilateral and non-lateralized ROIs
    in two columns (last ROI centered on the bottom row), alphabetically by display name.
    """
    # Load bilateral data
    bilateral_a = load_data(arm_dirs[ARM_A], effect_type, xname, 
                                 laterality_type='bilateral', correction_type=correction_type)
    bilateral_b = load_data(arm_dirs[ARM_B], effect_type, xname, 
                               laterality_type='bilateral', correction_type=correction_type)
    
    # Load non-lateralized data
    nonlat_a = load_data(arm_dirs[ARM_A], effect_type, xname, 
                              laterality_type='non_lateralized', correction_type=correction_type)
    nonlat_b = load_data(arm_dirs[ARM_B], effect_type, xname, 
                            laterality_type='non_lateralized', correction_type=correction_type)
    
    # Define the ROIs we want and their display names
    # Key: (data_key, source) → display_name
    selected_rois = sorted([
        ('Bilateral-Accumbens-area', 'bilateral', 'Accumbens'),
        ('Bilateral-ATR', 'bilateral', 'Anterior Radiations of the Thalamus'),
        ('Bilateral-Caudate', 'bilateral', 'Caudate'),
        ('Bilateral-CgC', 'bilateral', 'Cingulate Cingulum'),
        ('Fmin', 'non_lateralized', 'Forceps minor'),
        ('Bilateral-Pallidum', 'bilateral', 'Pallidum'),
        ('Bilateral-CgH', 'bilateral', 'Parahippocampal Cingulum'),
        ('Bilateral-Putamen', 'bilateral', 'Putamen'),
        ('Bilateral-Thalamus-Proper', 'bilateral', 'Thalamus'),
    ], key=lambda item: item[2])
    
    x_min, x_max = x_plot_min, x_plot_max
    
    fig = plt.figure(figsize=(15, 25))
    axes = _make_combined_roi_axes(fig, n_rois=len(selected_rois))
    
    for idx, (roi_key, source, disp_name) in enumerate(selected_rois):
        ax = axes[idx]
        
        # Select the right data source
        if source == 'bilateral':
            df_f = bilateral_a.get(roi_key)
            df_m = bilateral_b.get(roi_key)
        else:
            df_f = nonlat_a.get(roi_key)
            df_m = nonlat_b.get(roi_key)
        
        # Calculate y-axis limits
        y_min, y_max = float('inf'), float('-inf')
        if df_f is not None:
            df_f = df_f[(df_f['xvec'] >= x_min) & (df_f['xvec'] <= x_max)]
            y_min = min(y_min, df_f['CI_lower'].min())
            y_max = max(y_max, df_f['CI_upper'].max())
        if df_m is not None:
            df_m = df_m[(df_m['xvec'] >= x_min) & (df_m['xvec'] <= x_max)]
            y_min = min(y_min, df_m['CI_lower'].min())
            y_max = max(y_max, df_m['CI_upper'].max())
        
        # Plot ARM_A
        if df_f is not None:
            ax.plot(df_f['xvec'], df_f['mean'], color=colors[ARM_A], label=ARM_A)
            ax.fill_between(df_f['xvec'], df_f['CI_lower'], df_f['CI_upper'], 
                           color=ci_colors[ARM_A], alpha=0.3)
        
        # Plot ARM_B
        if df_m is not None:
            ax.plot(df_m['xvec'], df_m['mean'], color=colors[ARM_B], label=ARM_B)
            ax.fill_between(df_m['xvec'], df_m['CI_lower'], df_m['CI_upper'], 
                           color=ci_colors[ARM_B], alpha=0.3)
        
        ax.set_xlim([x_min, x_max])
        if y_min != float('inf'):
            ax.set_ylim([y_min, y_max])
        title_fs = 14 if len(disp_name) > 24 else 20
        ax.set_title(disp_name, fontsize=title_fs)
        ax.tick_params(axis='both', which='major', labelsize=20)
        ax.legend(loc="upper left", fontsize=18)
    
    # Add single shared axis labels for the whole figure
    fig.text(0.02, 0.5, f'Estimated {modality.upper()}', 
             va='center', rotation='vertical', fontsize=26)
    fig.text(0.5, 0.02, pheno_xlabel, 
             ha='center', fontsize=26)
    
    plt.tight_layout(rect=[0.05, 0.05, 1, 1])
    plt.show()
    
    save_dir = os.path.join(plots_base_directory, 'betas', 'combined_selected', correction_type, modality)
    base_name = f"{modality}_{effect_type}_{xname}_selected_ROIs_combined"
    
    if save_png or save_pdf or save_svg:
        os.makedirs(save_dir, exist_ok=True)
    
    if save_png:
        png_filename = os.path.join(save_dir, f"{base_name}.png")
        fig.savefig(png_filename, dpi=300, bbox_inches='tight')
        print(f"Saved combined PNG to {png_filename}")
    
    if save_pdf:
        pdf_filename = os.path.join(save_dir, f"{base_name}.pdf")
        fig.savefig(pdf_filename, dpi=300, bbox_inches='tight')
        print(f"Saved combined PDF to {pdf_filename}")
    
    if save_svg:
        svg_filename = os.path.join(save_dir, f"{base_name}.svg")
        fig.savefig(svg_filename, bbox_inches='tight')
        print(f"Saved combined SVG to {svg_filename}")

# Run it
for effect_type in effect_types:
    for xname in xnames:
        plot_combined_selected_rois(xname, effect_type, save_png=True, save_pdf=True, save_svg=True)
```

```python
# DERIVATIVES PLOTTING - Main (Lateralized and Non-lateralized)

# Helper function to load derivatives data
def load_derivatives_data(directory, effect_type, xname, laterality_type='lateralized', correction_type='fdr'):
    """
    Load derivatives data from the hierarchical directory structure.
    
    Parameters:
    -----------
    directory : str
        Base directory containing the design matrix results
    effect_type : str
        Type of effect (e.g., 'splines')
    xname : str
        Variable name (e.g., perc_median / adj_perc_median)
    laterality_type : str
        Type of laterality: 'lateralized', 'bilateral', or 'non_lateralized'
    correction_type : str
        Multiple comparison correction: 'bonferroni', 'fdr', or 'all_voxels'
    
    Returns:
    --------
    dict : Dictionary mapping ROI names to DataFrames
    """
    # Navigate to derivatives/laterality_type/correction_type subdirectory
    correction_dir = os.path.join(directory, 'derivatives', laterality_type, correction_type)
    if not os.path.exists(correction_dir):
        print(f"Warning: Directory not found: {correction_dir}")
        return {}
    
    pattern = f"{modality}_{effect_type}_{xname}_derivatives_CI_"
    files = [f for f in os.listdir(correction_dir) if f.startswith(pattern) and f.endswith('.txt')]
    
    print(f"\n{laterality_type} - Looking in: {correction_dir}")
    print(f"Pattern: {pattern}")
    print(f"Found {len(files)} files: {files}")
    
    data = {}
    for file in files:
        file_path = os.path.join(correction_dir, file)
        # Extract ROI name (everything after 'CI_' and before '.txt')
        roi_name = file.replace(pattern, '').replace('.txt', '')
        try:
            df = pd.read_csv(file_path)
            data[roi_name] = df
        except Exception as e:
            print(f"Error loading {file_path}: {e}")
    
    print(f"Loaded {len(data)} ROIs: {list(data.keys())}")
    return data

# Plot derivatives data
def plot_derivatives_data(data_a, data_b, xname, effect_type, laterality_type='lateralized', title_suffix="", save_pdf=False, save_png=False):
    # Check if there's any data to plot
    if not data_a and not data_b:
        print(f"No data found for {laterality_type} ROIs. Skipping plot.")
        return
    
    # Get all unique ROIs from both arms
    all_rois = set(list(data_a.keys()) + list(data_b.keys()))
    sorted_rois = sort_rois(all_rois)
    n_rois = len(sorted_rois)
    
    if n_rois == 0:
        print(f"No ROIs found for {laterality_type}. Skipping plot.")
        return
    
    fig, axes = plt.subplots(nrows=(n_rois + 1) // 2, ncols=2, figsize=(15, (n_rois + 1) * 2.5))

    # Larger font size for the main title
    fig.suptitle(f"{modality.upper()} {effect_type.capitalize()} Derivative for {xname.capitalize()} ({laterality_type}) {title_suffix}", fontsize=24)
    axes = axes.flatten()

    # Define the x-axis range limits
    x_min, x_max = x_plot_min, x_plot_max

    # Prepare PNG directory if saving individual ROI plots
    if save_png:
        png_dir = os.path.join(plots_base_directory, 'derivatives', laterality_type, correction_type, modality, 'pngs')
        os.makedirs(png_dir, exist_ok=True)

    for i in range(0, len(sorted_rois), 2):
        roi_pair = sorted_rois[i:i+2]  # Take pairs of ROIs (e.g., Left and Right)
        
        # Calculate combined y-axis limits for the pair
        y_min, y_max = float('inf'), float('-inf')
        for roi_name in roi_pair:
            if roi_name in data_a:
                df_a = data_a[roi_name]
                y_min = min(y_min, df_a['CI_lower'].min())
                y_max = max(y_max, df_a['CI_upper'].max())
            if roi_name in data_b:
                df_b = data_b[roi_name]
                y_min = min(y_min, df_b['CI_lower'].min())
                y_max = max(y_max, df_b['CI_upper'].max())
        
        # Skip this pair if no valid data found
        if y_min == float('inf') or y_max == float('-inf'):
            print(f"Warning: No data found for ROI pair {roi_pair}. Skipping.")
            continue

        for j, roi_name in enumerate(roi_pair):
            ax = axes[i + j]

            # Plot ARM_A within the x-axis range
            if roi_name in data_a:
                df_a = data_a[roi_name]
                df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                ax.plot(df_a['xvec'], df_a['mean_derivative'], color=colors[ARM_A], label=ARM_A)
                ax.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)

            # Plot ARM_B within the x-axis range
            if roi_name in data_b:
                df_b = data_b[roi_name]
                df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                ax.plot(df_b['xvec'], df_b['mean_derivative'], color=colors[ARM_B], label=ARM_B)
                ax.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)

            # Set axis limits and labels
            ax.set_xlim([x_min, x_max])
            ax.set_ylim([y_min, y_max])  # Set the same y-axis range for the pair

            # Increase font size for title, labels, and legend
            ax.set_title(display_name(roi_name), fontsize=20)
            ax.set_xlabel(pheno_xlabel, fontsize=20)
            ax.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=20)
            ax.tick_params(axis='both', which='major', labelsize=20)
            ax.legend(fontsize=18)

        # Save paired or unpaired ROIs as high-res PNG if requested
        if save_png:
            # Check if this is a left-right pair (handle both naming conventions)
            is_paired = False
            if len(roi_pair) == 2:
                # Check for "Left"/"Right" naming
                if (roi_pair[0].startswith("Left") and roi_pair[1].startswith("Right") and
                    roi_pair[0].replace("Left", "Right") == roi_pair[1]):
                    is_paired = True
                # Check for "LH-"/"RH-" naming (thalamus ROIs)
                elif (roi_pair[0].startswith("LH-") and roi_pair[1].startswith("RH-") and
                      roi_pair[0].replace("LH-", "RH-") == roi_pair[1]):
                    is_paired = True
            
            if is_paired:
                # Create a figure with 2 subplots for the pair
                fig_individual = plt.figure(figsize=(20, 6))
                
                for j, roi_name in enumerate(roi_pair):
                    ax_individual = fig_individual.add_subplot(1, 2, j + 1)
                    
                    # Plot ARM_A
                    if roi_name in data_a:
                        df_a = data_a[roi_name]
                        df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                        ax_individual.plot(df_a['xvec'], df_a['mean_derivative'], color=colors[ARM_A], label=ARM_A)
                        ax_individual.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)
                    
                    # Plot ARM_B
                    if roi_name in data_b:
                        df_b = data_b[roi_name]
                        df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                        ax_individual.plot(df_b['xvec'], df_b['mean_derivative'], color=colors[ARM_B], label=ARM_B)
                        ax_individual.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)
                    
                    # Set same limits and styling
                    ax_individual.set_xlim([x_min, x_max])
                    ax_individual.set_ylim([y_min, y_max])
                    ax_individual.set_title(display_name(roi_name), fontsize=20)
                    ax_individual.set_xlabel(pheno_xlabel, fontsize=20)
                    ax_individual.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=20)
                    ax_individual.tick_params(axis='both', which='major', labelsize=20)
                    ax_individual.legend(fontsize=18)
                
                # Save paired ROIs together
                base_roi_name = roi_pair[0].replace("Left", "").replace("LH-", "")
                png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_derivative_{base_roi_name}.png")
                fig_individual.tight_layout()
                fig_individual.savefig(png_filename, dpi=300)
                plt.close(fig_individual)
            else:
                # Save unpaired ROIs individually
                for roi_name in roi_pair:
                    fig_individual = plt.figure(figsize=(10, 6))
                    ax_individual = fig_individual.add_subplot(111)
                    
                    # Plot ARM_A
                    if roi_name in data_a:
                        df_a = data_a[roi_name]
                        df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
                        ax_individual.plot(df_a['xvec'], df_a['mean_derivative'], color=colors[ARM_A], label=ARM_A)
                        ax_individual.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], color=ci_colors[ARM_A], alpha=0.3)
                    
                    # Plot ARM_B
                    if roi_name in data_b:
                        df_b = data_b[roi_name]
                        df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
                        ax_individual.plot(df_b['xvec'], df_b['mean_derivative'], color=colors[ARM_B], label=ARM_B)
                        ax_individual.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], color=ci_colors[ARM_B], alpha=0.3)
                    
                    # Set same limits and styling
                    ax_individual.set_xlim([x_min, x_max])
                    ax_individual.set_ylim([y_min, y_max])
                    ax_individual.set_title(display_name(roi_name), fontsize=20)
                    ax_individual.set_xlabel(pheno_xlabel, fontsize=20)
                    ax_individual.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=20)
                    ax_individual.tick_params(axis='both', which='major', labelsize=20)
                    ax_individual.legend(fontsize=18)
                    
                    # Save as high-resolution PNG
                    png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_derivative_{roi_name}.png")
                    fig_individual.tight_layout()
                    fig_individual.savefig(png_filename, dpi=300)
                    plt.close(fig_individual)

    # Remove any unused axes
    for j in range(len(sorted_rois), len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])  # Adjust layout to fit the suptitle
    plt.show()  # Display

    # Save the figure as a PDF if requested
    if save_pdf:
        # Ensure the directory for PDFs exists
        pdf_dir = os.path.join(plots_base_directory, 'derivatives', laterality_type, correction_type, modality, 'pdfs')
        os.makedirs(pdf_dir, exist_ok=True)

        # Update filename to include modality and correction type
        pdf_filename = os.path.join(pdf_dir, f"{modality}_{effect_type}_{xname}_derivative_{title_suffix.replace(': ', '').replace(' ', '_')}.pdf")
        with PdfPages(pdf_filename) as pdf:
            pdf.savefig(fig)
        print(f"Saved plot as {pdf_filename}")

    # Save the combined figure (all ROIs) as a single PNG — same layout as the PDF
    if save_png:
        combined_png_dir = os.path.join(plots_base_directory, 'derivatives', laterality_type, correction_type, modality, 'pngs')
        os.makedirs(combined_png_dir, exist_ok=True)
        combined_png_filename = os.path.join(combined_png_dir, f"{modality}_{effect_type}_{xname}_derivative_all_ROIs_{title_suffix.replace(': ', '').replace(' ', '_')}.png")
        fig.savefig(combined_png_filename, dpi=300, bbox_inches='tight')
        print(f"Saved combined PNG to {combined_png_filename}")

    if save_png:
        print(f"Saved individual/paired ROI plots to {png_dir}")

# Load and plot derivatives data for the specified correction type
for effect_type in effect_types:
    for xname in xnames:
        # Plot lateralized ROIs
        data_a = load_derivatives_data(arm_dirs[ARM_A], effect_type, xname, laterality_type='lateralized', correction_type=correction_type)
        data_b = load_derivatives_data(arm_dirs[ARM_B], effect_type, xname, laterality_type='lateralized', correction_type=correction_type)
        plot_derivatives_data(data_a, data_b, xname, effect_type, laterality_type='lateralized',
                 title_suffix=f": {correction_type.replace('_', ' ').title()}", 
                 save_pdf=True, save_png=True)
        
        # Plot non-lateralized ROIs
        data_a = load_derivatives_data(arm_dirs[ARM_A], effect_type, xname, laterality_type='non_lateralized', correction_type=correction_type)
        data_b = load_derivatives_data(arm_dirs[ARM_B], effect_type, xname, laterality_type='non_lateralized', correction_type=correction_type)
        if data_a or data_b:  # Only plot if there's data
            plot_derivatives_data(data_a, data_b, xname, effect_type, laterality_type='non_lateralized',
                     title_suffix=f": {correction_type.replace('_', ' ').title()}", 
                     save_pdf=True, save_png=True)
```

```python
# DERIVATIVES PLOTTING - Bilateral ROIs

import matplotlib.cm as cm
import numpy as np

# Function to plot each bilateral ROI individually (noVE11 and VE11 on same graph) - DERIVATIVES
def plot_bilateral_derivatives_individual(data_a, data_b, xname, effect_type, title_suffix="", save_pdf=False, save_png=False):
    # Filter for bilateral ROIs only
    bilateral_rois_a = {k: v for k, v in data_a.items() if k.startswith('Bilateral-')}
    bilateral_rois_b = {k: v for k, v in data_b.items() if k.startswith('Bilateral-')}
    
    if not bilateral_rois_a and not bilateral_rois_b:
        print("No bilateral ROIs found in the data.")
        return
    
    # Get sorted list of bilateral ROIs
    all_bilateral_rois = sorted(set(list(bilateral_rois_a.keys()) + list(bilateral_rois_b.keys())))
    
    # Define x-axis range
    x_min, x_max = x_plot_min, x_plot_max
    
    # Create output directories
    if save_pdf:
        pdf_dir = os.path.join(plots_base_directory, 'derivatives', 'bilateral', correction_type, modality, 'bilateral_individual', 'pdfs')
        os.makedirs(pdf_dir, exist_ok=True)
    if save_png:
        png_dir = os.path.join(plots_base_directory, 'derivatives', 'bilateral', correction_type, modality, 'bilateral_individual', 'pngs')
        os.makedirs(png_dir, exist_ok=True)
    
    # Plot each bilateral ROI individually
    for roi_name in all_bilateral_rois:
        fig, ax = plt.subplots(1, 1, figsize=(10, 6))
        
        # Calculate y-axis limits for this ROI
        y_min, y_max = float('inf'), float('-inf')
        if roi_name in bilateral_rois_a:
            df = bilateral_rois_a[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            y_min = min(y_min, df['CI_lower'].min())
            y_max = max(y_max, df['CI_upper'].max())
        if roi_name in bilateral_rois_b:
            df = bilateral_rois_b[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            y_min = min(y_min, df['CI_lower'].min())
            y_max = max(y_max, df['CI_upper'].max())
        
        # Plot ARM_A
        if roi_name in bilateral_rois_a:
            df_a = bilateral_rois_a[roi_name]
            df_a = df_a[(df_a['xvec'] >= x_min) & (df_a['xvec'] <= x_max)]
            ax.plot(df_a['xvec'], df_a['mean_derivative'], color=colors[ARM_A], label=ARM_A, linewidth=2)
            ax.fill_between(df_a['xvec'], df_a['CI_lower'], df_a['CI_upper'], 
                           color=ci_colors[ARM_A], alpha=0.3)
        
        # Plot ARM_B
        if roi_name in bilateral_rois_b:
            df_b = bilateral_rois_b[roi_name]
            df_b = df_b[(df_b['xvec'] >= x_min) & (df_b['xvec'] <= x_max)]
            ax.plot(df_b['xvec'], df_b['mean_derivative'], color=colors[ARM_B], label=ARM_B, linewidth=2)
            ax.fill_between(df_b['xvec'], df_b['CI_lower'], df_b['CI_upper'], 
                           color=ci_colors[ARM_B], alpha=0.3)
        
        # Set axis limits and labels
        ax.set_xlim([x_min, x_max])
        ax.set_ylim([y_min, y_max])
        short_name = display_name(roi_name.replace('Bilateral-', ''))
        ax.set_title(f'{short_name}', fontsize=22)
        ax.set_xlabel(pheno_xlabel, fontsize=20)
        ax.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=20)
        ax.tick_params(axis='both', which='major', labelsize=18)
        ax.legend(fontsize=18)
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        # Save as PDF
        if save_pdf:
            pdf_filename = os.path.join(pdf_dir, f"{modality}_{effect_type}_{xname}_derivative_{roi_name}.pdf")
            fig.savefig(pdf_filename, dpi=300)
        
        # Save as PNG
        if save_png:
            png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_derivative_{roi_name}.png")
            fig.savefig(png_filename, dpi=300)
        
        # plt.close(fig)
    
    if save_pdf:
        print(f"Saved individual bilateral ROI derivative PDFs to {pdf_dir}")
    if save_png:
        print(f"Saved individual bilateral ROI derivative PNGs to {png_dir}")


# Function to plot selected bilateral ROIs together (one graph for VE11, one for noVE11) - DERIVATIVES
def plot_selected_bilateral_derivatives(data_a, data_b, selected_rois, xname, effect_type, 
                                  title_suffix="", save_pdf=False, save_png=False):
    """
    Plot selected bilateral ROI derivatives on the same graph.
    
    Parameters:
    -----------
    selected_rois : list
        List of bilateral ROI names to plot (e.g., ['Bilateral-Thalamus-Proper', 'Bilateral-Caudate'])
    """
    # Filter for selected bilateral ROIs
    bilateral_rois_a = {k: v for k, v in data_a.items() if k in selected_rois}
    bilateral_rois_b = {k: v for k, v in data_b.items() if k in selected_rois}
    
    if not bilateral_rois_a and not bilateral_rois_b:
        print("No selected bilateral ROIs found in the data.")
        return
    
    # Get sorted list of selected bilateral ROIs
    all_bilateral_rois = sorted(set(list(bilateral_rois_a.keys()) + list(bilateral_rois_b.keys())))
    n_rois = len(all_bilateral_rois)
    
    # Define x-axis range
    x_min, x_max = x_plot_min, x_plot_max
    
    # Calculate global y-axis limits across all selected bilateral ROIs for both arms
    y_min, y_max = float('inf'), float('-inf')
    for roi_name in all_bilateral_rois:
        if roi_name in bilateral_rois_a:
            df = bilateral_rois_a[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            y_min = min(y_min, df['CI_lower'].min())
            y_max = max(y_max, df['CI_upper'].max())
        if roi_name in bilateral_rois_b:
            df = bilateral_rois_b[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            y_min = min(y_min, df['CI_lower'].min())
            y_max = max(y_max, df['CI_upper'].max())
    
    # Generate distinct colors for each ROI
    # Use custom colors if available, otherwise fall back to colormap
    cmap = plt.cm.get_cmap('tab10', n_rois)
    roi_colors = {}
    for i, roi in enumerate(all_bilateral_rois):
        if roi in custom_roi_colors:
            roi_colors[roi] = custom_roi_colors[roi]
        else:
            roi_colors[roi] = cmap(i)
    
    # Create figure with two subplots (both arms)
    fig, (ax_a, ax_b) = plt.subplots(1, 2, figsize=(24, 8))
    
    # Plot ARM_A
    for roi_name in all_bilateral_rois:
        if roi_name in bilateral_rois_a:
            df = bilateral_rois_a[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            color = roi_colors[roi_name]
            
            # Extract short name (remove "Bilateral-" prefix for legend)
            short_name = display_name(roi_name.replace('Bilateral-', ''))
            
            ax_a.plot(df['xvec'], df['mean_derivative'], color=color, label=short_name, linewidth=2.5)
            ax_a.fill_between(df['xvec'], df['CI_lower'], df['CI_upper'], color=color, alpha=0.2)
    
    ax_a.set_xlim([x_min, x_max])
    ax_a.set_ylim([y_min, y_max])
    ax_a.set_title(f'{ARM_A} - Subcortical ROIs Derivatives', fontsize=24)
    ax_a.set_xlabel(pheno_xlabel, fontsize=22)
    ax_a.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=22)
    ax_a.tick_params(axis='both', which='major', labelsize=20)
    ax_a.grid(True, alpha=0.3)
    
    # Plot ARM_B
    for roi_name in all_bilateral_rois:
        if roi_name in bilateral_rois_b:
            df = bilateral_rois_b[roi_name]
            df = df[(df['xvec'] >= x_min) & (df['xvec'] <= x_max)]
            color = roi_colors[roi_name]
            
            # Extract short name (remove "Bilateral-" prefix for legend)
            short_name = display_name(roi_name.replace('Bilateral-', ''))
            
            ax_b.plot(df['xvec'], df['mean_derivative'], color=color, label=short_name, linewidth=2.5)
            ax_b.fill_between(df['xvec'], df['CI_lower'], df['CI_upper'], color=color, alpha=0.2)
    
    ax_b.set_xlim([x_min, x_max])
    ax_b.set_ylim([y_min, y_max])
    ax_b.set_title(f'{ARM_B} - Subcortical ROIs Derivatives', fontsize=24)
    ax_b.set_xlabel(pheno_xlabel, fontsize=22)
    ax_b.set_ylabel(f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', fontsize=22)
    ax_b.tick_params(axis='both', which='major', labelsize=20)
    ax_b.grid(True, alpha=0.3)
    
    # Main title
    fig.suptitle(f"{modality.upper()} {effect_type.capitalize()} Derivative for {xname.capitalize()} - Subcortical ROIs {title_suffix}", 
                 fontsize=24)
    
    # Create single shared legend at the bottom
    handles, labels = ax_a.get_legend_handles_labels()
    fig.legend(handles, labels, loc='lower center', bbox_to_anchor=(0.5, -0.02), 
               fontsize=20, ncol=min(len(labels), 5), frameon=True)
    
    plt.tight_layout(rect=[0, 0.08, 1, 0.97])
    plt.show()
    
    # Save as PDF if requested
    if save_pdf:
        pdf_dir = os.path.join(plots_base_directory, 'derivatives', 'bilateral', correction_type, modality, 'bilateral_combined', 'pdfs')
        os.makedirs(pdf_dir, exist_ok=True)
        pdf_filename = os.path.join(pdf_dir, f"{modality}_{effect_type}_{xname}_derivative_selected_bilateral_ROIs_{title_suffix.replace(': ', '').replace(' ', '_')}.pdf")
        fig.savefig(pdf_filename, dpi=300, bbox_inches='tight')
        print(f"Saved selected bilateral ROI derivative plot as {pdf_filename}")
    
    # Save as PNG if requested
    if save_png:
        png_dir = os.path.join(plots_base_directory, 'derivatives', 'bilateral', correction_type, modality, 'bilateral_combined', 'pngs')
        os.makedirs(png_dir, exist_ok=True)
        png_filename = os.path.join(png_dir, f"{modality}_{effect_type}_{xname}_derivative_selected_bilateral_ROIs.png")
        fig.savefig(png_filename, dpi=300, bbox_inches='tight')
        print(f"Saved selected bilateral ROI derivative plot as {png_filename}")


# Load data from bilateral directory and plot derivatives
for effect_type in effect_types:
    for xname in xnames:
        data_a = load_derivatives_data(arm_dirs[ARM_A], effect_type, xname, laterality_type='bilateral', correction_type=correction_type)
        data_b = load_derivatives_data(arm_dirs[ARM_B], effect_type, xname, laterality_type='bilateral', correction_type=correction_type)
        
        # Plot each bilateral ROI individually (noVE11 + VE11 together)
        plot_bilateral_derivatives_individual(data_a, data_b, xname, effect_type,
                                      title_suffix=f": {correction_type.replace('_', ' ').title()}",
                                      save_pdf=True, save_png=True)
        
        # Plot selected bilateral ROIs together
        selected_rois = [
            'Bilateral-Thalamus-Proper',
            'Bilateral-Caudate',
            'Bilateral-Putamen',
            'Bilateral-Accumbens-area',
            'Bilateral-Pallidum'
        ]
        
        plot_selected_bilateral_derivatives(data_a, data_b, selected_rois, xname, effect_type,
                                    title_suffix=f": {correction_type.replace('_', ' ').title()}",
                                    save_pdf=True, save_png=True)
```

```python
# Combined derivatives: selected bilateral + non-lateralized ROIs in a 3x2 grid (no CIs)

def plot_combined_selected_derivatives(xname, effect_type, save_png=False, save_pdf=False, save_svg=False):
    """
    Create a single combined figure with selected bilateral and non-lateralized ROI
    derivatives arranged in a 3x2 grid. Mean derivative lines only (no CI bands).
    """
    bilateral_a = load_derivatives_data(
        arm_dirs[ARM_A], effect_type, xname,
        laterality_type='bilateral', correction_type=correction_type,
    )
    bilateral_b = load_derivatives_data(
        arm_dirs[ARM_B], effect_type, xname,
        laterality_type='bilateral', correction_type=correction_type,
    )
    nonlat_a = load_derivatives_data(
        arm_dirs[ARM_A], effect_type, xname,
        laterality_type='non_lateralized', correction_type=correction_type,
    )
    nonlat_b = load_derivatives_data(
        arm_dirs[ARM_B], effect_type, xname,
        laterality_type='non_lateralized', correction_type=correction_type,
    )

    selected_rois = [
        ('Bilateral-Accumbens-area', 'bilateral', 'Accumbens'),
        ('Bilateral-Caudate', 'bilateral', 'Caudate'),
        ('Fmin', 'non_lateralized', 'Forceps minor'),
        ('Bilateral-Pallidum', 'bilateral', 'Pallidum'),
        ('Bilateral-Putamen', 'bilateral', 'Putamen'),
        ('Bilateral-Thalamus-Proper', 'bilateral', 'Thalamus'),
    ]

    x_min, x_max = x_plot_min, x_plot_max

    fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15, 15))
    axes = axes.flatten()

    for idx, (roi_key, source, disp_name) in enumerate(selected_rois):
        ax = axes[idx]

        if source == 'bilateral':
            df_f = bilateral_a.get(roi_key)
            df_m = bilateral_b.get(roi_key)
        else:
            df_f = nonlat_a.get(roi_key)
            df_m = nonlat_b.get(roi_key)

        y_min, y_max = float('inf'), float('-inf')
        if df_f is not None:
            df_f = df_f[(df_f['xvec'] >= x_min) & (df_f['xvec'] <= x_max)]
            y_min = min(y_min, df_f['mean_derivative'].min())
            y_max = max(y_max, df_f['mean_derivative'].max())
        if df_m is not None:
            df_m = df_m[(df_m['xvec'] >= x_min) & (df_m['xvec'] <= x_max)]
            y_min = min(y_min, df_m['mean_derivative'].min())
            y_max = max(y_max, df_m['mean_derivative'].max())

        if df_f is not None:
            ax.plot(df_f['xvec'], df_f['mean_derivative'], color=colors[ARM_A], label=ARM_A)
        if df_m is not None:
            ax.plot(df_m['xvec'], df_m['mean_derivative'], color=colors[ARM_B], label=ARM_B)

        ax.set_xlim([x_min, x_max])
        if y_min != float('inf'):
            ax.set_ylim([y_min, y_max])
        ax.set_title(disp_name, fontsize=20)
        ax.tick_params(axis='both', which='major', labelsize=20)
        ax.legend(fontsize=18)

    fig.text(0.02, 0.5, f'Change in estimated {modality.upper()} per unit {pheno_xlabel}', va='center', rotation='vertical', fontsize=26)
    fig.text(0.5, 0.02, pheno_xlabel, ha='center', fontsize=26)

    plt.tight_layout(rect=[0.05, 0.05, 1, 1])
    plt.show()

    save_dir = os.path.join(plots_base_directory, 'derivatives', 'combined_selected', correction_type, modality)
    base_name = f"{modality}_{effect_type}_{xname}_selected_ROIs_derivative_combined"

    if save_png or save_pdf or save_svg:
        os.makedirs(save_dir, exist_ok=True)

    if save_png:
        png_filename = os.path.join(save_dir, f"{base_name}.png")
        fig.savefig(png_filename, dpi=300, bbox_inches='tight')
        print(f"Saved combined derivative PNG to {png_filename}")

    if save_pdf:
        pdf_filename = os.path.join(save_dir, f"{base_name}.pdf")
        fig.savefig(pdf_filename, dpi=300, bbox_inches='tight')
        print(f"Saved combined derivative PDF to {pdf_filename}")

    if save_svg:
        svg_filename = os.path.join(save_dir, f"{base_name}.svg")
        fig.savefig(svg_filename, bbox_inches='tight')
        print(f"Saved combined derivative SVG to {svg_filename}")


for effect_type in effect_types:
    for xname in xnames:
        plot_combined_selected_derivatives(
            xname, effect_type, save_png=True, save_pdf=True, save_svg=True,
        )

```
